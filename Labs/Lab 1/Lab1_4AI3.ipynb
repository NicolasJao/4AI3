```python
A = np.zeros((3, 4))
print(A)

B = np.ones((3, 4))
print(B)

# Matrix addition
C = A + 2 * B
print(C)

print("--------")
# Matrix multiplication
A = np.array([[1,2,3], [4,5,6]])
B = np.array([[1,1], [2,2], [3,3]])
print(A)
print(B)
print(np.dot(A, B))

print("--------")
# Transpose
AT = np.transpose(A)
print(AT)
```
